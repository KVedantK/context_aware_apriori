```python
import json, warnings
from pathlib import Path
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
warnings.filterwarnings("ignore")
```

```python
MIN_SUPPORT, MIN_CONF = 0.02, 0.3

ds1 = json.load(open("ds_train1.json"))
ds2 = {d["item_id"]: d for d in json.load(open("ds2_new.json"))}
ID = {d["name"]: i for i, d in ds2.items()}
nm = lambda i: ds2[i]["name"]
```

```python
tx = [r["context"] + [r["outcome"]] for r in ds1]
outcomes = {r["outcome"] for r in ds1}
```

```python
te = TransactionEncoder(); df = pd.DataFrame(te.fit(tx).transform(tx), columns=te.columns_)
freq = apriori(df, min_support=MIN_SUPPORT, use_colnames=True)
```

```python
rules = association_rules(freq, num_itemsets=len(df), metric="confidence", min_threshold=MIN_CONF)
```

```python
print("Max antecedent length:", rules["antecedents"].apply(len).max())
```

```python
# Maximum frequent itemset size
print("Max frequent itemset:", freq["itemsets"].apply(len).max())

# Maximum antecedent size
print("Max antecedent:", rules["antecedents"].apply(len).max())

# Distribution of frequent itemset sizes
print(freq["itemsets"].apply(len).value_counts().sort_index())

# Distribution of antecedent sizes
print(rules["antecedents"].apply(len).value_counts().sort_index())
```
